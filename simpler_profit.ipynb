```python
import numpy as np
import pandas as pd 
import matplotlib as mpl
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.pipeline import Pipeline
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_absolute_error

df_train = pd.read_csv("../train_V2_cleaned.csv")
df_train = df_train.drop(columns=["outcome_damage_amount","outcome_damage_inc"])

df_train = df_train.sample(frac=1, random_state=304)

X = df_train.drop(columns=["outcome_profit"])
y = df_train["outcome_profit"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=304)

best_params = {'criterion': 'squared_error', 'learning_rate': 0.12, 'loss': 'squared_error', 'max_depth': 3, 'max_features': None, 'max_leaf_nodes': 54, 'min_samples_leaf': 3, 'min_samples_split': 14, 'n_estimators': 493, 'subsample': 0.87}

gbm_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("gbm", GradientBoostingRegressor(
        **best_params,
        random_state=304
    ))
])

gbm_pipeline.fit(X_train, y_train)

y_pred = gbm_pipeline.predict(X_test)

print("R²:", r2_score(y_test, y_pred))
print("MAE:", mean_absolute_error(y_test, y_pred))
```

# Train on all data

```python
df_score = pd.read_csv("../score_cleaned.csv")
df_result = df_score.copy()
df_score = df_score.drop(columns=['outcome_damage_inc', 'outcome_damage_inc_proba'], errors='ignore')

gbm_pipeline.fit(X, y)
y_score_pred = gbm_pipeline.predict(df_score)

df_result['outcome_profit'] = y_score_pred
df_result.to_csv('../score_cleaned.csv', index=False)
```
