```python
import pandas as pd
```

```python
df = pd.read_csv('score_cleaned.csv')
```

```python
def get_rank(df, top=150):
    df['net_profit'] = df['outcome_profit'] - 1.2*df['outcome_damage_amount']
    ranking = df['net_profit']
    sorted_ranking = ranking.sort_values(ascending=False)
    if (top > sorted_ranking.size):
        top = sorted_ranking.size
    sorted_ranking = sorted_ranking[:top]
    return sorted_ranking
```

```python
indexes = get_rank(df).index
df_output = df.iloc[indexes]
df_output.to_csv('score_selected_observations.csv')
```
